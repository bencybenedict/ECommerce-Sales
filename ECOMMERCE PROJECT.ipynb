# 4 VISUALIZATION

*4.1 Visualization for Product count by each category*

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.countplot(data=df,x='product_category',color='green')
plt.xlabel('product_category')
plt.ylabel('product_count')
plt.title('product_count by category')
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```

Summary

Electronics (1777) records the highest product count, followed by Clothing (1531), indicating stronger product availability in these categories.Beauty (723) has the lowest product count, while Home (969) maintains a moderate product presence across the dataset.

*4.2 Visualization for Orders vs Customer rating*

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.histplot(df['customer_rating'],kde=True , bins = 10,color= 'blue')
plt.title('Customer Rating')
plt.show()
```

Summary

The customer ratings range from 1 to 5, with the highest concentration around the mid-range ratings (approximately 2–4). Overall, the distribution indicates moderate and varied customer satisfaction, with fewer customers giving the highest rating of 5, highlighting potential opportunities to improve customer experience.


*4.3 Visualization for Average Revenue by each year*

```python
sns.boxplot(x='revenue',y='product_category',data=df,color='c')
plt.show()
```

Summary

Revenue is fairly similar across all product categories, with median revenue around the same level. Several high-revenue outliers are visible, especially in Electronics and Clothing, indicating that a small number of orders generate significantly higher revenue than typical orders.


*4.4 Visualization for Payment Methods*

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.countplot(data=df,x='payment_method',color='#2E8B57')
plt.xlabel('payment mathod')
plt.ylabel('count')
plt.title('Payment Method')
plt.xticks(rotation=45)
plt.show()
```

Summary

Card is the most widely used payment method, followed by Cash on Delivery (COD). Wallet has the lowest usage, indicating that customers strongly prefer card and COD for making payments.

*4.5 Visualization of Delivery Days taken for Orders*

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
sns.countplot(data=df,x='delivery_days',color=(0.5,0.4,0.8))
plt.xlabel('delivery days')
plt.ylabel('count')
plt.title('Number of Orders by Delivery Days')
plt.xticks()
plt.show()
```

Summary

The number of orders is fairly consistent across 1–11 delivery days, with only small variations. The highest number of orders delivered in 10 days, while 3 days has the lowest, indicating that orders are distributed relatively evenly across different delivery durations.

*4.6 Visualization of orders by each Category*


```python
counts= df['product_category'].value_counts().head(4)
import matplotlib.pyplot as plt

plt.figure(figsize = (7,6))

plt.pie(
    counts,
    labels = counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Product Category Distribution")
plt.tight_layout()
plt.show()
```

Summary

Electronics has the largest share at 35.5%, followed by Clothing at 30.6%. Home accounts for 19.4%, while Beauty has the smallest share at 14.5%, showing that customers have a stronger preference for Electronics and Clothing products.

*4.7 Correlation between the numeric columns of the dataset*

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize = (8,7))
corr = df.corr(numeric_only=True)
sns.heatmap(corr,annot=True, cmap='Set2', linewidths=0.5)
plt.title('Correlation between the numeric columns of the dataset')
plt.show()
```

*4.8 Visualization for relationship between unit price and revenue*

```python
sns.scatterplot(
    data = df,
    x = 'unit_price',
    y = 'revenue',
    hue = 'quantity'
)
```

Summary

The analysis shows a strong positive relationship between unit price and revenue, with revenue generally increasing as unit price rises. Higher quantities contribute to greater revenue, making both price and quantity important factors influencing overall sales revenue.

*4.9 Visualization for Customer Rating by Product Category*

```python
plt.figure(figsize=(15,6))

sns.countplot(data=df, x='product_category', hue='customer_rating')

plt.title('Customer Rating by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Number of Products')

plt.show()
```

Summary

The graph shows that customer ratings vary across different product categories.
It helps compare the number of products receiving each rating within each category.
